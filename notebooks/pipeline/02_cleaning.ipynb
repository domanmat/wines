# Cleaning — Bronze → Silver

Loads the raw Bronze parquet (all strings), applies the canonical type
casts and per-column cleaning, and writes `wine_reviews_silver.parquet`.

The encoding + XGBoost cells at the bottom are an **exploratory tail** —
they will move to their own notebooks when Sprint 4 (features) and
Sprint 7 (modeling) start.

```python
import numpy as np
import pandas as pd
import itables
from itables import show

itables.options.columnDefs = [{"className": "dt-left", "targets": "_all"}]

BRONZE_PATH = r"..\..\.data\wine_reviews_bronze.parquet"

df = pd.read_parquet(BRONZE_PATH)
print(f"Bronze shape: {df.shape}")
df.dtypes
```

# Cleaning and casting

## Casting column types

```python
cast_columns = [
    ("alcohol",         "float"),
    ("vintage",         "int"),
    ("case_production", "int"),
    ("retail",          "float"),
    ("rating",          "int"),
    ("bottle_size",     "float"),
]

df_cast = df.copy()

for column, dtype in cast_columns:
    nulls_before = df_cast[column].isna().sum()
    cast_col = pd.to_numeric(df_cast[column], errors="coerce")
    nulls_after = cast_col.isna().sum()
    error_mask = df_cast[column].notna() & cast_col.isna()
    failures = error_mask.sum()

    df_cast[f"{column}_cast"] = cast_col
    df_cast[f"{column}_error"] = df_cast[column].where(error_mask)

    print(f"{column:20s}  nulls before={nulls_before:>6,}  after={nulls_after:>6,}  failures={failures:>6,}")
    if failures > 0:
        print(df_cast.loc[error_mask, column].unique())
```

### alcohol

```python
def fix_alcohol(v):
    if pd.isna(v):
        return np.nan
    if v > 25.0:
        return np.nan
    if v > 7.0:
        return v / 100
    if v > 2.5:
        return np.nan
    if v > 0.7:
        return v / 10
    if v > 0.25:
        return np.nan
    if v >= 0.05:
        return v
    return np.nan

df_cast["alcohol"] = df_cast["alcohol_cast"].map(fix_alcohol)
```

### bottle_size

```python
show(df_cast["bottle_size"].value_counts(dropna=False))
```

```python
def parse_bottle_size(s):
    if pd.isna(s):
        return np.nan
    s = str(s).strip()
    s_lower = s.lower()
    if s_lower.endswith("l ml"):
        try:
            return float(s[:-4].strip()) * 1000
        except ValueError:
            return np.nan
    elif s_lower.endswith("ml"):
        try:
            return float(s[:-2].strip())
        except ValueError:
            return np.nan
    elif s_lower.endswith("l"):
        try:
            return float(s[:-1].strip()) * 1000
        except ValueError:
            return np.nan
    else:
        try:
            return float(s)
        except ValueError:
            return np.nan

df_cast["bottle_size"] = df_cast["bottle_size"].map(parse_bottle_size)
```

```python
valid_sizes = {750.0, 375.0, 500.0, 1500.0, 3000.0, 250.0, 1000.0, 187.0, 620.0}

def clean_bottle_size(v):
    if pd.isna(v):
        return np.nan
    if v == 75.0:
        return 750.0
    if v == 751.0:
        return 750.0
    if v == 740.0:
        return 750.0
    if v in valid_sizes:
        return v
    return np.nan

df_cast["bottle_size"] = df_cast["bottle_size"].map(clean_bottle_size)
```

## Deduplication

Drop exact-row duplicates and repeated `(slug, date_of_review)` pairs.
Both are accidental — same wine + same review date = same record.
More aggressive keys (e.g. `(name, vintage)`) would merge legitimate sibling reviews, so we stop here.

```python
before = len(df_cast)

# Exact-row duplicates
df_cast = df_cast.drop_duplicates().reset_index(drop=True)
after_exact = len(df_cast)

# (slug, date_of_review) duplicates
df_cast = df_cast.drop_duplicates(subset=["slug", "date_of_review"]).reset_index(drop=True)
after_slug = len(df_cast)

print(f"Exact duplicates removed:           {before - after_exact:,}")
print(f"(slug, date) duplicates removed:    {after_exact - after_slug:,}")
print(f"Rows remaining:                     {after_slug:,}")
```

## Save as Parquet (Silver)

```python
SILVER_PATH = r"..\..\.data\wine_reviews_silver.parquet"

# Silver = original columns replaced with cleaned typed versions + is_nv flag.
# _cast / _error / _ord helper columns stay out of Silver.
silver_cols = list(df.columns) + ["is_nv"]
df_silver = df_cast[silver_cols].copy()

# Stable primary key so derived feature tables (keywords, embeddings, ...)
# can join back reliably — slug is NOT unique (42 dupes).
df_silver.insert(0, "wine_id", range(len(df_silver)))

df_silver.to_parquet(SILVER_PATH, index=False)
print(f"Silver shape: {df_silver.shape}")
print(f"Silver written to {SILVER_PATH}")
```
